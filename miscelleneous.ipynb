```python

class Employee:
    def __init__(self):
        self.name = "Anu"
        self._salary = 50000
        self.__password = "12345"

    def show_parent(self):
      print("name:",self.name)
      print("salary:",self._salary)
      print("password",self.__password)


class Manager(Employee):
  def show_manager(self):
    print("name:",self.name)
    print("salary",self._salary)


manager = Manager()
manager.show_parent()
manager.show_manager()
```
